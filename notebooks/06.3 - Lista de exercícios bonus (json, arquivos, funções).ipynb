## Exercício 1 — Sistema de notas com funções

Uma escola armazenou as notas dos alunos no arquivo `notas.json`. Cada aluno possui uma lista de notas.

Crie um programa que leia esse arquivo e gere `boletim.json`, contendo a média e a situação de cada aluno.

O programa deve usar funções auxiliares. Por exemplo:

- uma função para calcular a média de uma lista de notas;
- uma função para definir a situação do aluno a partir da média;
- opcionalmente, uma função geral para montar o boletim completo.

Regras:

- média maior ou igual a `7`: `"aprovado"`;
- média maior ou igual a `5` e menor que `7`: `"recuperacao"`;
- média menor que `5`: `"reprovado"`.

Formato esperado:

```json
[
  {"aluno": "Ana", "media": 8.17, "situacao": "aprovado"}
]
```

### Preparação dos arquivos

````python
import json

dados = [
    {"aluno": "Ana", "notas": [8.0, 7.5, 9.0]},
    {"aluno": "Bruno", "notas": [5.0, 6.0, 6.5]},
    {"aluno": "Carla", "notas": [9.0, 9.5, 10.0]},
    {"aluno": "Daniel", "notas": [4.0, 5.5, 6.0]}
]

with open("notas.json", "w", encoding="utf-8") as f:
    json.dump(dados, f, indent=2, ensure_ascii=False)

print("Arquivo notas.json criado.")
````

### Gabarito

````python
import json

def calcular_media(notas):
    soma = 0

    for nota in notas:
        soma += nota

    return soma / len(notas)


def definir_situacao(media):
    if media >= 7:
        return "aprovado"
    elif media >= 5:
        return "recuperacao"
    else:
        return "reprovado"


def gerar_boletim(alunos):
    boletim = []

    for aluno in alunos:
        media = calcular_media(aluno["notas"])
        situacao = definir_situacao(media)

        boletim.append({
            "aluno": aluno["aluno"],
            "media": round(media, 2),
            "situacao": situacao
        })

    return boletim


with open("notas.json", "r", encoding="utf-8") as f:
    alunos = json.load(f)

boletim = gerar_boletim(alunos)

with open("boletim.json", "w", encoding="utf-8") as f:
    json.dump(boletim, f, indent=2, ensure_ascii=False)

print("Arquivo boletim.json criado.")
````

## Exercício 2 — Controle de biblioteca

Uma biblioteca possui um cadastro de livros em JSON e um arquivo CSV com os empréstimos realizados.

Crie um programa que cruze os dois arquivos e gere `relatorio_biblioteca.json`, com a quantidade de empréstimos de cada livro.

O programa deve:

- ler os livros;
- criar um dicionário usando o id do livro como chave;
- ler o arquivo CSV manualmente, usando `split(",")`;
- contar quantas vezes cada livro foi emprestado;
- gerar o relatório final em JSON.

Formato esperado:

```json
[
  {"titulo": "Dom Casmurro", "autor": "Machado de Assis", "emprestimos": 3}
]
```

### Preparação dos arquivos

````python
import json

livros = [
    {"id": 1, "titulo": "Dom Casmurro", "autor": "Machado de Assis"},
    {"id": 2, "titulo": "Capitaes da Areia", "autor": "Jorge Amado"},
    {"id": 3, "titulo": "A Hora da Estrela", "autor": "Clarice Lispector"}
]

with open("livros.json", "w", encoding="utf-8") as f:
    json.dump(livros, f, indent=2, ensure_ascii=False)

conteudo_csv = '''id_livro,usuario
1,Ana
2,Bruno
1,Carla
3,Ana
1,Daniel'''

with open("emprestimos.csv", "w", encoding="utf-8") as f:
    f.write(conteudo_csv)

print("Arquivos livros.json e emprestimos.csv criados.")
````

### Gabarito

````python
import json

with open("livros.json", "r", encoding="utf-8") as f:
    livros = json.load(f)

livros_por_id = {}

for livro in livros:
    id_livro = str(livro["id"])

    livros_por_id[id_livro] = {
        "titulo": livro["titulo"],
        "autor": livro["autor"],
        "emprestimos": 0
    }

with open("emprestimos.csv", "r", encoding="utf-8") as f:
    linhas = f.readlines()

for linha in linhas[1:]:
    linha = linha.strip()
    partes = linha.split(",")

    id_livro = partes[0]

    livros_por_id[id_livro]["emprestimos"] += 1

relatorio = []

for id_livro in livros_por_id:
    relatorio.append({
        "titulo": livros_por_id[id_livro]["titulo"],
        "autor": livros_por_id[id_livro]["autor"],
        "emprestimos": livros_por_id[id_livro]["emprestimos"]
    })

with open("relatorio_biblioteca.json", "w", encoding="utf-8") as f:
    json.dump(relatorio, f, indent=2, ensure_ascii=False)

print("Arquivo relatorio_biblioteca.json criado.")
````

## Exercício 3 — Carrinho de compras com desconto

Uma loja virtual salvou os itens de um carrinho no arquivo `carrinho.json`.

Crie um programa que gere `resumo_carrinho.json`, contendo:

- subtotal de cada item;
- total bruto da compra;
- desconto aplicado;
- total final.

O programa deve usar funções auxiliares. Por exemplo:

- uma função para calcular o subtotal de um item;
- uma função para calcular o desconto;
- opcionalmente, uma função geral para montar o resumo.

Regra:

- se o total bruto for maior que `1000`, aplicar 10% de desconto;
- caso contrário, o desconto deve ser `0`.

Formato esperado:

```json
{
  "itens": [
    {"produto": "Mouse", "subtotal": 160.0}
  ],
  "total_bruto": 1210.0,
  "desconto": 121.0,
  "total_final": 1089.0
}
```

### Preparação dos arquivos

````python
import json

carrinho = [
    {"produto": "Mouse", "preco": 80.0, "quantidade": 2},
    {"produto": "Teclado", "preco": 150.0, "quantidade": 1},
    {"produto": "Monitor", "preco": 900.0, "quantidade": 1}
]

with open("carrinho.json", "w", encoding="utf-8") as f:
    json.dump(carrinho, f, indent=2, ensure_ascii=False)

print("Arquivo carrinho.json criado.")
````

### Gabarito

````python
import json

def calcular_subtotal(preco, quantidade):
    return preco * quantidade


def calcular_desconto(total):
    if total > 1000:
        return total * 0.10
    else:
        return 0


def gerar_resumo(carrinho):
    itens_resumo = []
    total_bruto = 0

    for item in carrinho:
        subtotal = calcular_subtotal(item["preco"], item["quantidade"])
        total_bruto += subtotal

        itens_resumo.append({
            "produto": item["produto"],
            "subtotal": round(subtotal, 2)
        })

    desconto = calcular_desconto(total_bruto)
    total_final = total_bruto - desconto

    resumo = {
        "itens": itens_resumo,
        "total_bruto": round(total_bruto, 2),
        "desconto": round(desconto, 2),
        "total_final": round(total_final, 2)
    }

    return resumo


with open("carrinho.json", "r", encoding="utf-8") as f:
    carrinho = json.load(f)

resumo = gerar_resumo(carrinho)

with open("resumo_carrinho.json", "w", encoding="utf-8") as f:
    json.dump(resumo, f, indent=2, ensure_ascii=False)

print("Arquivo resumo_carrinho.json criado.")
````

## Exercício 4 — Agenda de eventos

Você recebeu um arquivo `eventos.txt`, em que cada linha representa uma inscrição em um evento.

Cada linha segue o formato:

```txt
evento;data;participante
```

Um mesmo evento pode aparecer em várias linhas, com participantes diferentes.

Crie um programa que leia o arquivo e gere `eventos.json`, agrupando os participantes por evento.

O programa deve:

- ler cada linha;
- separar os campos usando `split(";")`;
- criar um dicionário em que cada evento seja uma chave;
- armazenar a data do evento;
- armazenar a lista de participantes;
- gerar uma lista final de eventos em JSON.

Formato esperado:

```json
[
  {
    "evento": "Python para Dados",
    "data": "2026-05-10",
    "participantes": ["Ana", "Carla"],
    "total_participantes": 2
  }
]
```

### Preparação dos arquivos

````python
conteudo = '''Python para Dados;2026-05-10;Ana
Banco de Dados;2026-05-12;Bruno
Python para Dados;2026-05-10;Carla
Machine Learning;2026-05-15;Daniel
Banco de Dados;2026-05-12;Ana'''

with open("eventos.txt", "w", encoding="utf-8") as f:
    f.write(conteudo)

print("Arquivo eventos.txt criado.")
````

### Gabarito

````python
import json

eventos = {}

with open("eventos.txt", "r", encoding="utf-8") as f:
    linhas = f.readlines()

for linha in linhas:
    linha = linha.strip()
    partes = linha.split(";")

    nome_evento = partes[0]
    data = partes[1]
    participante = partes[2]

    if nome_evento not in eventos:
        eventos[nome_evento] = {
            "data": data,
            "participantes": []
        }

    eventos[nome_evento]["participantes"].append(participante)

resultado = []

for nome_evento in eventos:
    participantes = eventos[nome_evento]["participantes"]

    resultado.append({
        "evento": nome_evento,
        "data": eventos[nome_evento]["data"],
        "participantes": participantes,
        "total_participantes": len(participantes)
    })

with open("eventos.json", "w", encoding="utf-8") as f:
    json.dump(resultado, f, indent=2, ensure_ascii=False)

print("Arquivo eventos.json criado.")
````

## Exercício 5 — Mini sistema de inventário

Uma empresa registrou entradas e saídas de produtos no arquivo `movimentacoes.json`.

Cada movimentação possui:

- produto;
- tipo da movimentação;
- quantidade.

Crie um programa que calcule o estoque final de cada produto e gere `estoque_final.json`.

Regras:

- `"entrada"` aumenta o estoque;
- `"saida"` diminui o estoque;
- se o produto ainda não existir no dicionário de estoque, ele deve começar com valor `0`.

Formato esperado:

```json
[
  {"produto": "Mouse", "estoque": 7},
  {"produto": "Teclado", "estoque": 3},
  {"produto": "Monitor", "estoque": 4}
]
```

### Preparação dos arquivos

````python
import json

movimentacoes = [
    {"produto": "Mouse", "tipo": "entrada", "quantidade": 10},
    {"produto": "Teclado", "tipo": "entrada", "quantidade": 5},
    {"produto": "Mouse", "tipo": "saida", "quantidade": 3},
    {"produto": "Monitor", "tipo": "entrada", "quantidade": 4},
    {"produto": "Teclado", "tipo": "saida", "quantidade": 2}
]

with open("movimentacoes.json", "w", encoding="utf-8") as f:
    json.dump(movimentacoes, f, indent=2, ensure_ascii=False)

print("Arquivo movimentacoes.json criado.")
````

### Gabarito

````python
import json

with open("movimentacoes.json", "r", encoding="utf-8") as f:
    movimentacoes = json.load(f)

estoque = {}

for mov in movimentacoes:
    produto = mov["produto"]
    tipo = mov["tipo"]
    quantidade = mov["quantidade"]

    if produto not in estoque:
        estoque[produto] = 0

    if tipo == "entrada":
        estoque[produto] += quantidade
    elif tipo == "saida":
        estoque[produto] -= quantidade

resultado = []

for produto in estoque:
    resultado.append({
        "produto": produto,
        "estoque": estoque[produto]
    })

with open("estoque_final.json", "w", encoding="utf-8") as f:
    json.dump(resultado, f, indent=2, ensure_ascii=False)

print("Arquivo estoque_final.json criado.")
````
