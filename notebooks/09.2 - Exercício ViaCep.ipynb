# Projeto Final

Desenvolva um programa que:

1. Consulte uma IA.
2. Solicite um prefixo de CEP adequado.
3. Extraia o prefixo retornado.
4. Gere 100 CEPs.
5. Consulte a API ViaCEP.
6. Utilize try/except.
7. Aguarde um pequeno intervalo entre consultas.
8. Identifique CEPs válidos e inválidos.
9. Salve tudo em JSON.
10. Informe estatísticas ao final.

---

## Gabarito Completo


````python
from google import genai
import requests
import json
import time

client = genai.Client(
    api_key="SUA_CHAVE"
)

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=(
        "Informe apenas os 5 primeiros dígitos "
        "de uma região do Rio de Janeiro que "
        "possua muitos CEPs válidos. "
        "Retorne apenas os 5 dígitos."
    )
)

prefixo = response.text.strip()

print("Prefixo sugerido:", prefixo)

resultados = []

validos = 0
invalidos = 0

for i in range(100):

    cep = prefixo + f"{i:03d}"

    try:

        resp = requests.get(
            f"https://viacep.com.br/ws/{cep}/json/",
            timeout=5
        )

        resp.raise_for_status()

        dados = resp.json()

        if "erro" in dados:

            invalidos += 1

            resultados.append({
                "cep": cep,
                "status": "invalido"
            })

        else:

            validos += 1

            resultados.append({
                "cep": cep,
                "status": "valido",
                "dados": dados
            })

    except Exception as e:

        resultados.append({
            "cep": cep,
            "status": "erro",
            "mensagem": str(e)
        })

    time.sleep(0.2)

with open(
    "ceps_ia.json",
    "w",
    encoding="utf-8"
) as arquivo:

    json.dump(
        resultados,
        arquivo,
        ensure_ascii=False,
        indent=4
    )

print("Consulta finalizada")
print("Válidos:", validos)
print("Inválidos:", invalidos)
````
