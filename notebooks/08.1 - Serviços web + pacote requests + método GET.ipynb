## 1. A ideia central: cliente e servidor

Grande parte da internet funciona com uma lógica de **cliente e servidor**.

O **cliente** é quem faz uma solicitação.

Exemplos de clientes:

- Um navegador, como Chrome, Firefox ou Edge.
- Um aplicativo de celular.
- Um script Python.
- Um notebook Jupyter.

O **servidor** é quem recebe a solicitação, processa o pedido e devolve uma resposta.

Exemplos de servidores:

- Um site.
- Uma API de clima.
- Uma API de mapas.
- Uma API de inteligência artificial.
- Um sistema de cadastro de usuários.

Podemos imaginar o fluxo assim:

```text
Cliente
(Python, navegador, app)

        ↓ requisição HTTP

Servidor
(site, API, sistema Web)

        ↓ resposta HTTP

Cliente
(recebe texto, HTML, JSON, imagem etc.)
```

Nesta aula, nosso Python será o **cliente**.

## 2. O que é HTTP?

HTTP significa **HyperText Transfer Protocol**.

É o protocolo usado para comunicação entre clientes e servidores na Web.

Quando você acessa um site, por exemplo:

```text
https://www.google.com
```

o navegador está fazendo uma requisição HTTP para um servidor.

Quando usamos Python com a biblioteca `requests`, fazemos algo parecido:

```python
requests.get("https://alguma-api.com/dados")
```

Ou seja:

> O Python envia uma requisição pela internet e recebe uma resposta.

## 3. Métodos HTTP principais

As requisições HTTP podem usar diferentes **métodos**.

Cada método indica a intenção da requisição.

| Método | Ideia principal | Exemplo de uso |
|---|---|---|
| `GET` | Buscar dados | Buscar uma lista de produtos |
| `POST` | Enviar/criar dados | Enviar um formulário de cadastro |
| `PUT` | Atualizar dados inteiros | Atualizar todos os dados de um usuário |
| `PATCH` | Atualizar parte dos dados | Alterar apenas o e-mail de um usuário |
| `DELETE` | Remover dados | Excluir um registro |

Nesta primeira aula, vamos focar principalmente no método **GET**.

O método `GET` é usado quando queremos **buscar informações**.

## 4. O que é uma API?

API significa **Application Programming Interface**.

Em termos práticos, uma API é uma forma de um programa conversar com outro programa.

Exemplo:

```text
Python → API de clima → retorna temperatura
Python → API de filmes → retorna lista de filmes
Python → API de IA → retorna uma resposta gerada por modelo
Python → API pública de testes → retorna dados fictícios
```

## 7. Primeira requisição GET

Vamos fazer nossa primeira requisição GET.

Usaremos o seguinte endereço:

```text
https://jsonplaceholder.typicode.com/posts/1
```

Esse endpoint retorna um único post fictício.

````python
url = "https://jsonplaceholder.typicode.com/posts/1"

resposta = requests.get(url)

resposta
````

O objeto retornado é um objeto do tipo `Response`.

Ele representa a resposta HTTP recebida do servidor.

Quando aparece algo como:

```text
<Response [200]>
```

significa que a requisição foi feita e o servidor respondeu com o código de status `200`.

O código `200` geralmente significa sucesso.

## 8. Verificando o status da resposta

Toda resposta HTTP possui um **código de status**.

Esse código informa se a requisição funcionou ou se houve algum problema.

````python
resposta.status_code
````

Alguns códigos comuns:

| Código | Significado |
|---|---|
| 200 | OK — requisição bem-sucedida |
| 201 | Created — recurso criado com sucesso |
| 400 | Bad Request — erro na requisição |
| 401 | Unauthorized — não autorizado |
| 403 | Forbidden — acesso proibido |
| 404 | Not Found — recurso não encontrado |
| 500 | Internal Server Error — erro no servidor |

Em APIs reais, verificar o status é uma etapa importante antes de tentar usar os dados.

## 12. O que é JSON?

JSON significa **JavaScript Object Notation**.

Apesar do nome, ele é usado por várias linguagens, inclusive Python.

Um JSON se parece muito com dicionários e listas em Python.

Exemplo de JSON:

```json
{
  "nome": "Ana",
  "idade": 25,
  "aprovada": true
}
```

Em Python, isso se parece com:

```python
{
    "nome": "Ana",
    "idade": 25,
    "aprovada": True
}
```

Diferenças importantes:

| JSON | Python |
|---|---|
| `true` | `True` |
| `false` | `False` |
| `null` | `None` |

Quando usamos `.json()`, o `requests` converte a resposta JSON para estruturas Python.

## 18. Usando `params` com requests

Podem informar, por exemplo:

## 21. Cuidado: nem toda resposta é JSON

Se a resposta não for JSON, o método `.json()` pode gerar erro.

Por isso, em programas mais robustos, é comum verificar o `Content-Type`.

````python
content_type = resposta.headers["Content-Type"]

content_type
````

````python
if "application/json" in content_type:
    dados = resposta.json()
    print("Resposta JSON convertida com sucesso.")
else:
    print("A resposta não parece ser JSON.")
````

## 22. Tratando erros com `try` e `except`

Requisições Web podem falhar por vários motivos:

- falta de internet;
- servidor fora do ar;
- URL incorreta;
- tempo de resposta muito longo;
- bloqueio de acesso;
- erro interno do servidor.

Por isso, é importante tratar erros.

````python
try:
    resposta = requests.get("https://jsonplaceholder.typicode.com/posts/1")
    print(resposta.status_code)

except Exception as erro:
    print("Ocorreu um erro:")
    print(erro)
````

## 23. Usando timeout

Um bom hábito é usar `timeout`.

O `timeout` define quantos segundos o Python deve esperar antes de desistir da requisição.

Sem timeout, um programa pode ficar travado esperando uma resposta.

````python
try:
    resposta = requests.get(
        "https://jsonplaceholder.typicode.com/posts/1",
        timeout=5
    )

    print(resposta.status_code)

except Exception as erro:
    print("Erro na requisição:")
    print(erro)
````

## 24. Criando uma função para buscar dados

Agora vamos encapsular a lógica em uma função.

A função abaixo busca um post pelo ID.

````python
def buscar_post(id_post):
    url = f"https://jsonplaceholder.typicode.com/posts/{id_post}"

    resposta = requests.get(url, timeout=5)

    if resposta.status_code == 200:
        return resposta.json()
    else:
        return None
````

````python
post = buscar_post(3)

post
````

````python
if post is not None:
    print(post["title"])
else:
    print("Post não encontrado.")
````

## 26. Mini projeto guiado: relatório simples de usuários

Vamos criar um pequeno programa que:

1. Busca todos os usuários.
2. Conta quantos usuários existem.
3. Mostra nome, e-mail e cidade de cada usuário.

Esse exemplo é interessante porque a resposta da API vira uma lista de dicionários, e os alunos precisam navegar pela estrutura.

````python
url = "https://jsonplaceholder.typicode.com/users"

resposta = requests.get(url, timeout=5)

usuarios = resposta.json()

print(f"Total de usuários: {len(usuarios)}")
print()

for usuario in usuarios:
    nome = usuario["name"]
    email = usuario["email"]
    cidade = usuario["address"]["city"]

    print(f"Nome: {nome}")
    print(f"E-mail: {email}")
    print(f"Cidade: {cidade}")
    print("-" * 40)
````

Observe que a cidade está dentro de um dicionário aninhado:

```python
usuario["address"]["city"]
```

Isso é muito comum em respostas JSON de APIs reais.

## 27. Mini projeto guiado: posts por usuário

Agora vamos combinar duas informações:

- usuários;
- posts.

Objetivo:

Para cada usuário, mostrar quantos posts ele possui.

````python
usuarios = requests.get(
    "https://jsonplaceholder.typicode.com/users",
    timeout=5
).json()

posts = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    timeout=5
).json()

for usuario in usuarios:
    id_usuario = usuario["id"]
    nome = usuario["name"]

    posts_do_usuario = [
        post for post in posts
        if post["userId"] == id_usuario
    ]

    print(f"{nome}: {len(posts_do_usuario)} posts")
````

Esse exemplo já se aproxima de uma análise de dados simples:

- coleta dados de uma API;
- transforma JSON em listas e dicionários;
- filtra registros;
- calcula uma métrica;
- exibe um resumo.

## 28. Mini projeto guiado: função de relatório

Vamos transformar o relatório anterior em uma função.

````python
def relatorio_posts_por_usuario():
    usuarios = requests.get(
        "https://jsonplaceholder.typicode.com/users",
        timeout=5
    ).json()

    posts = requests.get(
        "https://jsonplaceholder.typicode.com/posts",
        timeout=5
    ).json()

    relatorio = []

    for usuario in usuarios:
        id_usuario = usuario["id"]

        posts_do_usuario = [
            post for post in posts
            if post["userId"] == id_usuario
        ]

        item = {
            "id_usuario": id_usuario,
            "nome": usuario["name"],
            "email": usuario["email"],
            "quantidade_posts": len(posts_do_usuario)
        }

        relatorio.append(item)

    return relatorio
````

````python
relatorio = relatorio_posts_por_usuario()

relatorio[:3]
````

Agora temos uma lista de dicionários com dados resumidos.

Isso pode ser usado depois para:

- criar um DataFrame;
- salvar em CSV;
- alimentar um dashboard;
- enviar para outra API;
- usar em uma aplicação Flask.

## 30. Boas práticas ao consumir APIs

1. Verifique o status code antes de usar a resposta.

Nesta aula usamos uma API pública simples, mas APIs reais podem exigir:

- autenticação;
- token;
- chave de API;

## Gabarito — Exercício 8

````python
def buscar_titulos_posts_usuario(id_usuario):
    parametros = {
        "userId": id_usuario
    }

    resposta = requests.get(
        "https://jsonplaceholder.typicode.com/posts",
        params=parametros
    )

    if resposta.status_code == 200:
        posts = resposta.json()

        titulos = []

        for post in posts:
            titulos.append(post["title"])

        return titulos
    else:
        return []


titulos = buscar_titulos_posts_usuario(1)

titulos
````

Uma versão usando list comprehension:

````python
def buscar_titulos_posts_usuario(id_usuario):
    parametros = {
        "userId": id_usuario
    }

    resposta = requests.get(
        "https://jsonplaceholder.typicode.com/posts",
        params=parametros
    )

    if resposta.status_code == 200:
        posts = resposta.json()
        return [post["title"] for post in posts]
    else:
        return []


buscar_titulos_posts_usuario(1)
````

## Gabarito — Exercício 10

````python
usuarios = requests.get(
    "https://jsonplaceholder.typicode.com/users"
).json()

posts = requests.get(
    "https://jsonplaceholder.typicode.com/posts"
).json()

relatorio = []

for usuario in usuarios:
    id_usuario = usuario["id"]

    quantidade = 0

    for post in posts:
        if post["userId"] == id_usuario:
            quantidade += 1

    item = {
        "nome": usuario["name"],
        "email": usuario["email"],
        "quantidade_posts": quantidade
    }

    relatorio.append(item)

relatorio
````

Uma versão alternativa usando list comprehension:

````python
usuarios = requests.get(
    "https://jsonplaceholder.typicode.com/users"
).json()

posts = requests.get(
    "https://jsonplaceholder.typicode.com/posts"
).json()

relatorio = []

for usuario in usuarios:
    id_usuario = usuario["id"]

    posts_do_usuario = [
        post for post in posts
        if post["userId"] == id_usuario
    ]

    relatorio.append({
        "nome": usuario["name"],
        "email": usuario["email"],
        "quantidade_posts": len(posts_do_usuario)
    })

relatorio
````
