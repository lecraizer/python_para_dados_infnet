
# Importante: uma única aplicação Flask

Durante toda a aplicação normalmente criamos apenas um objeto Flask:

```python
app = Flask(__name__)
```

Esse objeto representa toda a aplicação.

Após criar o objeto `app`, adicionamos quantas rotas quisermos.

Somente depois iniciamos o servidor com:

```python
app.run(debug=True)
```

Fluxo recomendado:

1. Criar a aplicação.
2. Definir as rotas.
3. Executar `app.run()` apenas uma vez.


````python
from flask import Flask, jsonify

app = Flask(__name__)
````


# Primeira rota

Uma rota representa um endereço da aplicação.


````python
@app.route("/")
def home():
    return "Servidor funcionando!"
````


# Múltiplas rotas

Podemos associar diferentes URLs a diferentes funções.


````python
@app.route("/aluno")
def aluno():
    return jsonify({
        "nome": "Ana",
        "nota": 9.5,
        "curso": "Python"
    })

@app.route("/curso")
def curso():
    return jsonify({
        "nome": "Python para Dados",
        "carga_horaria": 40
    })
````


# Rotas com parâmetros

Uma das funcionalidades mais úteis do Flask é capturar informações diretamente da URL.

Exemplo:

```text
/quadrado/5
```

O valor 5 será recebido pela função Python.


````python
@app.route("/quadrado/<int:numero>")
def quadrado(numero):

    return jsonify({
        "numero": numero,
        "quadrado": numero ** 2
    })
````


Teste:

```text
http://127.0.0.1:5000/quadrado/5
```

Resultado:

```json
{
    "numero": 5,
    "quadrado": 25
}
```



# Múltiplos parâmetros

Podemos receber mais de um valor pela URL.


````python
@app.route("/soma/<int:a>/<int:b>")
def soma(a, b):

    return jsonify({
        "a": a,
        "b": b,
        "resultado": a + b
    })
````


Exemplo:

```text
/soma/10/20
```



# Trabalhando com listas Python

APIs normalmente retornam coleções de dados.


````python
filmes = [
    {"id": 1, "titulo": "Interestelar"},
    {"id": 2, "titulo": "Parasita"},
    {"id": 3, "titulo": "Whiplash"}
]

@app.route("/filmes")
def listar_filmes():
    return jsonify(filmes)
````


# Buscando um elemento pelo ID

Esse é um padrão extremamente comum em APIs.


````python
@app.route("/filme/<int:id>")
def buscar_filme(id):

    for filme in filmes:
        if filme["id"] == id:
            return jsonify(filme)

    return jsonify({
        "erro": "Filme não encontrado"
    })
````


Exemplos:

```text
/filme/1
/filme/2
/filme/3
```
