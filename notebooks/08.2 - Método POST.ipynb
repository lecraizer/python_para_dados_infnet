# 1. GET vs POST

## GET

Usado principalmente para consultar informações.

```python
requests.get(...)
```

Exemplo:

```python
requests.get(
    "https://viacep.com.br/ws/22041001/json/"
)
```

## POST

Usado quando enviamos dados para o servidor processar.

```python
requests.post(...)
```

Exemplos:

- Login
- Cadastro de usuário
- Envio de formulário
- Upload de arquivo
- APIs de IA

---

Importante:

Tanto GET quanto POST retornam respostas.

```python
resp = requests.get(...)
resp.json()
```

e

```python
resp = requests.post(...)
resp.json()
```

são igualmente válidos.

# 3. Headers

Headers são metadados da requisição.



## Content-Type

Informa o formato do body.

```python
headers = {
    "Content-Type": "application/json"
}
```

Outros exemplos:

```text
application/json
text/plain
text/html
application/xml
multipart/form-data
```



## Accept

Informa o formato que gostaríamos de receber.

```python
headers = {
    "Accept": "application/json"
}
```



## User-Agent

Informa quem está fazendo a requisição.

Exemplo simples:

```python
headers = {
    "User-Agent": "Mozilla/5.0"
}
```

Exemplo mais realista:

```python
headers = {
    "User-Agent":
    "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 Chrome/138.0 Safari/537.36"
}
```

Muitos sites usam User-Agent para detectar navegadores.



## Authorization

Usado para autenticação.

Exemplo:

```python
headers = {
    "Authorization": "Bearer MINHA_CHAVE"
}
```

Muito comum em APIs.



# 4. O que é o Body?

O body contém os dados principais da operação.

Exemplo:

```python
body = {
    "nome": "Luis",
    "idade": 37
}
```

O formato do body é definido pela documentação da API.

Não existe um formato universal.



# 5. Exemplo de POST Tradicional


````python
import requests

url = "https://httpbin.org/post"

headers = {
    "Content-Type": "application/json"
}

body = {
    "nome": "Luis",
    "idade": 37
}

resposta = requests.post(
    url,
    headers=headers,
    json=body
)

dados = resposta.json()
dados
````


# 6. json=body

Quando fazemos:

```python
requests.post(
    url,
    json=body
)
```

O requests converte automaticamente o dicionário Python para JSON.

Exemplo:

```python
body = {
    "nome": "Luis"
}
```

vira:

```json
{
    "nome": "Luis"
}
```



# 7. O que o Servidor Espera?

Imagine uma API que espera:

```json
{
    "usuario": "...",
    "senha": "..."
}
```

O servidor pode acessar:

```python
dados["usuario"]
dados["senha"]
```

Se você enviar:

```json
{
    "banana": 123
}
```

provavelmente receberá erro.



# 8. Exemplo de Login

Login geralmente utiliza POST.

Por quê?

Porque estamos enviando dados ao servidor.

```python
body = {
    "usuario": "admin",
    "senha": "123"
}
```


````python
import requests

url = "https://httpbin.org/post"

headers = {
    "Content-Type": "application/json"
}

body = {
    "usuario": "admin",
    "senha": "123"
}

resposta = requests.post(
    url,
    headers=headers,
    json=body
)

dados = resposta.json()
dados
````


# 9. Respostas

Após o POST:

```python
resposta = requests.post(...)
```

podemos acessar:

```python
resposta.status_code
resposta.headers
resposta.text
resposta.json()
```


````python

print(resposta.status_code)
print(resposta.text)
print(resposta.headers)
````


# 10. Códigos de Status

```text

400 -> requisição inválida

403 -> proibido
404 -> não encontrado
500 -> erro interno do servidor
```



# 11. Timeout

Evita que o programa fique esperando para sempre.

```python
requests.post(
    url,
    json=body,
    timeout=5
)
```



# 12. Tratamento de Erros


````python

try:

    resposta = requests.post(
        url,
        json=body,
        timeout=5
    )

    print(resposta.status_code)

except Exception as erro:

    print("Erro:", erro)
````


# 13. Exemplo Completo

Exemplo genérico reunindo tudo.


````python
import requests

url = "https://httpbin.org/post"

headers = {
    "Content-Type": "application/json",
    "Accept": "application/json",
    "User-Agent": "Mozilla/5.0"
}

body = {
    "mensagem": "Olá servidor"
}

try:

    resposta = requests.post(
        url,
        headers=headers,
        json=body,
        timeout=5
    )

    print("Status:", resposta.status_code)

    dados = resposta.json()

    print("\nBody recebido pelo servidor:")
    print(dados["json"])

    print("\nUser-Agent recebido:")
    print(dados["headers"]["User-Agent"])

except Exception as erro:

    print("Erro:", erro)
````
