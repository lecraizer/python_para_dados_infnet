
# Exercício Prático – API de Biblioteca com Flask

Neste exercício, você deverá criar uma API simples utilizando Flask.

A ideia é simular uma pequena biblioteca que disponibiliza informações sobre livros através de rotas.

O objetivo é praticar:

- criação de uma aplicação Flask;
- criação de rotas;
- retorno de dados em JSON;
- uso de listas e dicionários;
- uso de parâmetros na URL;
- busca de informações usando `for` e `if`.



# Contexto

Uma biblioteca deseja disponibilizar seus livros através de uma API.

A API deve permitir:

1. acessar uma página inicial;
2. listar todos os livros cadastrados;
3. buscar um livro específico pelo seu `id`;
4. buscar livros de um determinado autor.

Neste exercício, os dados serão armazenados diretamente em uma lista Python.



# Parte 1 – Criando a aplicação

Crie uma aplicação Flask.

A aplicação deve ter uma rota principal:

```text
/
```

Essa rota deve retornar a mensagem:

```text
Bem-vindo à API da Biblioteca!
```



# Parte 2 – Criando os dados

Crie uma lista chamada `livros` contendo os seguintes livros:

| id | título | autor | ano |
|---:|--------|-------|----:|
| 1 | Dom Casmurro | Machado de Assis | 1899 |
| 2 | O Hobbit | J. R. R. Tolkien | 1937 |
| 3 | 1984 | George Orwell | 1949 |
| 4 | O Pequeno Príncipe | Antoine de Saint-Exupéry | 1943 |
| 5 | Cem Anos de Solidão | Gabriel García Márquez | 1967 |

Cada livro deve ser representado por um dicionário Python.



# Parte 3 – Listando todos os livros

Crie a rota:

```text
/livros
```

Essa rota deve retornar todos os livros cadastrados.

O retorno deve ser feito usando `jsonify()`.



# Parte 4 – Buscando livro por ID

Crie a rota:

```text
/livro/<id>
```

Essa rota deve receber um número inteiro pela URL.

Exemplos de teste:

```text
/livro/1
/livro/3
/livro/5
```

A rota deve retornar apenas o livro solicitado.



# Parte 5 – Tratando livro inexistente

Caso o usuário acesse um ID que não existe, por exemplo:

```text
/livro/100
```

a API deve retornar:

```json
{
    "erro": "Livro não encontrado"
}
```



# Parte 6 – Desafio extra: buscar por autor

Crie a rota:

```text
/autor/<nome>
```

Essa rota deve retornar todos os livros escritos pelo autor informado.

Exemplo:

```text
/autor/George Orwell
```

Retorno esperado:

```json
[
    {
        "id": 3,
        "titulo": "1984",
        "autor": "George Orwell",
        "ano": 1949
    }
]
```

Dica: compare os nomes usando `.lower()` para evitar problemas com letras maiúsculas e minúsculas.



# Observação importante

Crie a aplicação Flask apenas uma vez:

```python
app = Flask(__name__)
```

Depois disso, adicione todas as rotas ao mesmo objeto `app`.

O comando `app.run()` deve aparecer apenas uma vez, no final do código.



# Espaço para resolução

Use a célula abaixo para desenvolver sua solução.



# Gabarito


````python
from flask import Flask
from flask import jsonify

app = Flask(__name__)

livros = [
    {
        "id": 1,
        "titulo": "Dom Casmurro",
        "autor": "Machado de Assis",
        "ano": 1899
    },
    {
        "id": 2,
        "titulo": "O Hobbit",
        "autor": "J. R. R. Tolkien",
        "ano": 1937
    },
    {
        "id": 3,
        "titulo": "1984",
        "autor": "George Orwell",
        "ano": 1949
    },
    {
        "id": 4,
        "titulo": "O Pequeno Príncipe",
        "autor": "Antoine de Saint-Exupéry",
        "ano": 1943
    },
    {
        "id": 5,
        "titulo": "Cem Anos de Solidão",
        "autor": "Gabriel García Márquez",
        "ano": 1967
    }
]
````

````python
# @app.route("/")
# def home():
#     return "Bem-vindo à API da Biblioteca!"


@app.route("/")
def home():

    return """
    <h1>API da Biblioteca</h1>

    <p>Bem-vindo à API da Biblioteca!</p>

    <h2>Rotas disponíveis</h2>

    <ul>
        <li><a href="/livros">/livros</a></li>
        <li>/livro/&lt;id&gt;</li>
        <li>/autor/&lt;nome&gt;</li>
    </ul>

    <h2>Exemplos</h2>

    <ul>
        <li><a href="/livro/1">/livro/1</a></li>
        <li><a href="/livro/3">/livro/3</a></li>
        <li><a href="/autor/George Orwell">/autor/George Orwell</a></li>
    </ul>
    """

@app.route("/livros")
def listar_livros():
    return jsonify(livros)


# import json
# from flask import Response

# @app.route("/livros")
# def livros():

#     return Response(
#         json.dumps(lista_livros, indent=4, ensure_ascii=False),
#         mimetype="application/json"
#     )


@app.route("/livro/<int:id>")
def buscar_livro(id):

    for livro in livros:
        if livro["id"] == id:
            return jsonify(livro)

    return jsonify({
        "erro": "Livro não encontrado"
    })




@app.route("/autor/<nome>")
def buscar_autor(nome):

    resultado = []

    for livro in livros:
        if livro["autor"].lower() == nome.lower():
            resultado.append(livro)

    return jsonify(resultado)


app.run(debug=False, port=5001)
````
