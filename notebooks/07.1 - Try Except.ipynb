# Tratamento de Exceções em Python

## `try`, `except`, `else` e `finally`

Nesta aula, vamos estudar como lidar com erros em Python usando tratamento de exceções.

A ideia central é simples:

> Existem partes do programa que podem dar erro durante a execução.  
> Em vez de deixar o programa quebrar de forma inesperada, podemos capturar esse erro e decidir o que fazer.

Isso é muito comum em situações como:

- converter texto em número;
- dividir valores;
- acessar posições em listas;
- acessar chaves em dicionários;
- abrir arquivos;
- ler arquivos JSON;
- receber entradas digitadas pelo usuário;
- processar dados inconsistentes.

# 1. O que é uma exceção?

Uma **exceção** é um erro que acontece durante a execução do programa.

Exemplo:

```python
10 / 0
```

Esse código gera um erro chamado `ZeroDivisionError`.

Outro exemplo:

```python
int("abc")
```

Esse código gera um erro chamado `ValueError`.

Em Python, quando um erro acontece e não é tratado, o programa para e mostra uma mensagem técnica chamada **traceback**.

````python
10 / 0
````

O erro acima não é um erro de sintaxe.  
O código está escrito de forma válida, mas a operação é impossível.

Isso é diferente de algo como:

```python
print("Olá"
```

Nesse caso, existe um erro de sintaxe: o código nem consegue começar a rodar corretamente.

# 2. Erro de sintaxe vs erro de execução

É importante separar duas ideias:

| Tipo de erro | Quando acontece | Exemplo |
|---|---|---|
| Erro de sintaxe | Antes do programa rodar | esquecer parênteses, dois pontos, aspas |
| Erro de execução | Durante a execução | dividir por zero, converter `"abc"` para inteiro |

O `try/except` serve para lidar principalmente com **erros de execução**.

# 3. A estrutura básica do `try/except`

A estrutura mais simples é:

```python
try:
    # código que pode dar erro
except TipoDoErro:
    # o que fazer se esse erro acontecer
```

A ideia é:

1. Python tenta executar o bloco `try`;
2. se nenhum erro acontecer, o `except` é ignorado;
3. se o erro esperado acontecer, Python pula para o bloco `except`;
4. o programa continua rodando depois disso.

````python
try:
    numerador = int(input("Digite o numerador: "))
    denominador = int(input("Digite o denominador: "))

    resultado = numerador / denominador

    print(f"O resultado da divisão é: {resultado}")

except ZeroDivisionError:
    print("Erro: divisão por zero não é permitida.")
````

No exemplo acima, estamos tratando apenas `ZeroDivisionError`.

Isso significa que o programa sabe lidar com divisão por zero, mas ainda pode quebrar se o usuário digitar uma entrada inválida, como `"abc"`.

# 4. Tratando mais de um tipo de erro

Um mesmo bloco `try` pode gerar erros diferentes.

Por exemplo:

```python
int(input(...))
```

pode gerar `ValueError`.

Já:

```python
numerador / denominador
```

pode gerar `ZeroDivisionError`.

Podemos tratar os dois separadamente.

````python
try:
    numerador = int(input("Digite o numerador: "))
    denominador = int(input("Digite o denominador: "))

    resultado = numerador / denominador

    print(f"O resultado da divisão é: {resultado}")

except ValueError:
    print("Erro: você precisa digitar números inteiros.")

except ZeroDivisionError:
    print("Erro: divisão por zero não é permitida.")
````

Essa é uma prática melhor do que usar apenas um `except` genérico, porque cada erro recebe uma mensagem mais adequada.

Compare:

```python
except:
    print("Deu erro")
```

com:

```python
except ValueError:
    print("Digite apenas números inteiros.")

except ZeroDivisionError:
    print("Não é possível dividir por zero.")
```

A segunda versão ajuda muito mais o usuário.

# 5. `ValueError`

O `ValueError` acontece quando o tipo da operação até faz sentido, mas o valor não é adequado.

Exemplo clássico:

```python
int("abc")
```

A função `int()` sabe converter strings para inteiros, mas `"abc"` não representa um número inteiro.

````python
try:
    numero = int(input("Digite um número inteiro: "))
    print(f"Você digitou o número: {numero}")

except ValueError:
    print("Erro: isso não é um número inteiro válido.")
````

````python
try:
    int("abc")
except ValueError:
    print("Erro: valor inválido!")
````

Outros exemplos que podem gerar `ValueError`:

```python
float("texto")
int("3.5")
```

O segundo caso é interessante: `"3.5"` parece número, mas não é um inteiro válido para `int()` diretamente.

````python
valores = ["10", "3.5", "abc"]

for valor in valores:
    try:
        numero = int(valor)
        print(f"{valor} convertido para {numero}")
    except ValueError:
        print(f"Não foi possível converter {valor!r} para inteiro.")
````

# 6. `ZeroDivisionError`

O `ZeroDivisionError` acontece quando tentamos dividir por zero.

Essa exceção é muito comum em programas que fazem cálculos.

````python
try:
    a = 10
    b = 0

    resultado = a / b

    print(resultado)

except ZeroDivisionError:
    print("Erro: não é possível dividir por zero.")
````

Em muitos casos, também poderíamos evitar esse erro usando `if`.

Exemplo:

```python
if b != 0:
    resultado = a / b
else:
    print("Não é possível dividir por zero.")
```

Então surge uma dúvida importante:

> Quando usar `if` e quando usar `try/except`?

Em geral:

- use `if` quando a condição é simples e esperada;
- use `try/except` quando a operação pode falhar por vários motivos ou quando o erro é mais naturalmente capturado durante a execução.

# 7. `TypeError`

O `TypeError` acontece quando tentamos fazer uma operação com tipos incompatíveis.

Exemplo:

```python
"2" + 2
```

Python não sabe somar uma string com um número inteiro.

````python
try:
    resultado = "2" + 2
    print(resultado)

except TypeError:
    print("Erro: tipos incompatíveis!")
````

Outro exemplo:

```python
len(10)
```

A função `len()` espera algo com tamanho, como uma string, lista ou dicionário.  
Um número inteiro não possui tamanho.

````python
try:
    tamanho = len(10)
    print(tamanho)

except TypeError:
    print("Erro: esse tipo de dado não possui tamanho.")
````

# 8. `IndexError`

O `IndexError` acontece quando tentamos acessar uma posição inexistente em uma lista.

Lembre:

```python
lista = [10, 20, 30]
```

Os índices válidos são:

```python
0, 1, 2
```

Se tentarmos acessar `lista[5]`, o índice não existe.

````python
try:
    lista = [10, 20, 30]
    print(lista[5])

except IndexError:
    print("Erro: índice fora do alcance!")
````

Esse erro é muito comum em loops, especialmente quando trabalhamos com a posição atual e a próxima posição.

````python
lista = [10, 20, 30, 40, 50]

for i in range(len(lista)):
    try:
        print(lista[i] + lista[i + 1])
    except IndexError:
        print("A lista chegou ao fim.")
````

O exemplo acima funciona, mas deve ser analisado com cuidado.

Ele usa o erro como uma forma de detectar o final da lista.  
Isso pode ser útil em alguns contextos, mas nem sempre é a solução mais clara.

# 9. `KeyError`

O `KeyError` acontece quando tentamos acessar uma chave inexistente em um dicionário.

Exemplo:

````python
try:
    dicionario = {"a": 1, "b": 2}
    print(dicionario["c"])

except KeyError:
    print("Erro: chave não encontrada!")
````

Uma alternativa ao `try/except` é verificar antes se a chave existe:

````python
dicionario = {"a": 1, "b": 2}

chave = "c"

if chave in dicionario:
    print(dicionario[chave])
else:
    print("Chave não encontrada.")
````

Outra alternativa é usar o método `.get()`:

````python
dicionario = {"a": 1, "b": 2}

print(dicionario.get("c", "Valor padrão"))
````

Então, para dicionários, temos três possibilidades comuns:

```python
dicionario[chave]       # pode gerar KeyError
dicionario.get(chave)   # retorna None se não encontrar
chave in dicionario     # testa antes
```

Cada uma pode fazer sentido em um contexto diferente.

# 10. `FileNotFoundError`

O `FileNotFoundError` acontece quando tentamos abrir um arquivo que não existe.

Esse erro é muito comum ao trabalhar com arquivos.

````python
try:
    arquivo = open("arquivo_inexistente.txt", "r", encoding="utf-8")
    conteudo = arquivo.read()
    arquivo.close()

    print(conteudo)

except FileNotFoundError:
    print("Erro: arquivo não encontrado.")
````

A forma mais recomendada de abrir arquivos em Python é usando `with`.

Com `with`, o arquivo é fechado automaticamente ao final do bloco.

````python
try:
    with open("arquivo_inexistente.txt", "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()

    print(conteudo)

except FileNotFoundError:
    print("Erro: arquivo não encontrado.")
````

# 11. Erros ao trabalhar com JSON

Ao trabalhar com JSON, dois erros comuns são:

- o arquivo não existir;
- o arquivo existir, mas estar mal formatado.

Para ler JSON, usamos o módulo `json`.

````python
import json

try:
    with open("dados.json", "r", encoding="utf-8") as arquivo:
        dados = json.load(arquivo)

    print(dados)

except FileNotFoundError:
    print("Erro: arquivo JSON não encontrado.")

except json.JSONDecodeError:
    print("Erro: o arquivo existe, mas não contém um JSON válido.")
````

Esse exemplo é muito importante, porque mostra que o `try/except` não serve apenas para erros matemáticos.

Ele também aparece muito em:

- leitura de arquivos;
- APIs;
- dados externos;
- entrada de usuário;
- processamento de informações que podem estar incompletas ou corrompidas.

# 12. Usando vários `except`

Podemos usar quantos blocos `except` forem necessários.

A ideia é tratar cada erro de forma específica.

````python
try:
    numero1 = float(input("Digite o primeiro número: "))
    numero2 = float(input("Digite o segundo número: "))

    resultado = numero1 / numero2

    print("Resultado:", resultado)

except ValueError:
    print("Erro: digite apenas números.")

except ZeroDivisionError:
    print("Erro: não é possível dividir por zero.")

except Exception as erro:
    print("Ocorreu um erro inesperado:", erro)
````

O último bloco:

```python
except Exception as erro:
```

captura erros mais gerais.

Ele pode ser útil, mas deve ser usado com cuidado.  
Se usado de qualquer forma, pode esconder problemas importantes do código.

# 13. Capturando a mensagem do erro

Às vezes, queremos guardar o erro em uma variável para entender melhor o que aconteceu.

Exemplo:

````python
try:
    numero = int("abc")

except ValueError as erro:
    print("Tipo do erro:", type(erro))
    print("Mensagem original:", erro)
````

Isso é útil principalmente durante a depuração ou para criar logs.

Mas, para o usuário final, geralmente mostramos uma mensagem mais amigável.

# 14. `else` no tratamento de exceções

O `else` pode ser usado junto com `try/except`.

Ele roda apenas se **nenhum erro acontecer** no bloco `try`.

Estrutura:

```python
try:
    código que pode dar erro
except AlgumErro:
    tratamento do erro
else:
    código executado se não houve erro
```

````python
try:
    numero = int(input("Digite um número inteiro: "))

except ValueError:
    print("Erro: valor inválido.")

else:
    print("Conversão realizada com sucesso.")
    print("Número digitado:", numero)
````

O `else` não é obrigatório.

Muitas vezes, conseguimos escrever o código sem ele.  
Mas ele ajuda a separar:

- o código que pode dar erro;
- o código que só deve rodar se tudo deu certo.

# 15. `finally`

O `finally` é um bloco que executa sempre:

- se deu erro;
- se não deu erro;
- se o erro foi tratado.

Estrutura:

```python
try:
    código
except AlgumErro:
    tratamento
finally:
    código que sempre roda
```

Ele é útil principalmente para ações de encerramento, como:

- fechar arquivos;
- encerrar conexões;
- liberar recursos;
- imprimir uma mensagem final;
- limpar algum estado temporário.

No início do curso, ele não é o mais importante.  
O essencial é entender bem `try` e `except`.  
Mas vale conhecer a existência do `finally`.

````python
try:
    numero = int(input("Digite um número inteiro: "))
    print("Número:", numero)

except ValueError:
    print("Erro: valor inválido.")

finally:
    print("Fim da tentativa.")
````

Com arquivos, o `finally` pode ser usado para garantir fechamento.  
Porém, em Python moderno, geralmente preferimos usar `with`, que faz isso automaticamente.

# 16. Usar erro como teste: solução prática ou gambiarra?

Às vezes, programadores usam `try/except` como uma forma de testar se algo funciona.

Exemplo:

````python
lista = [10, 20, 30, 40, 50]

for i in range(len(lista)):
    try:
        print(lista[i] + lista[i + 1])
    except IndexError:
        print("Lista chegou ao fim.")
````

Esse código funciona.

Ele tenta acessar o próximo elemento da lista.  
Quando chega no último elemento, `lista[i + 1]` não existe e gera `IndexError`.

O `except` captura esse erro e mostra:

```python
Lista chegou ao fim.
```

Isso é uma espécie de “teste pela tentativa”.

Em inglês, existe uma ideia comum em Python chamada:

> **EAFP — Easier to Ask Forgiveness than Permission**

Em tradução livre:

> É mais fácil pedir perdão do que pedir permissão.

Ou seja, em vez de verificar tudo antes, você tenta fazer a operação e trata o erro se ele acontecer.

# 17. Comparando duas abordagens

## Abordagem 1 — Tentando e tratando o erro

```python
for i in range(len(lista)):
    try:
        print(lista[i] + lista[i + 1])
    except IndexError:
        print("Lista chegou ao fim.")
```

## Abordagem 2 — Evitando o erro com `range`

```python
for i in range(len(lista) - 1):
    print(lista[i] + lista[i + 1])
```

A segunda abordagem é mais simples nesse caso.

````python
lista = [10, 20, 30, 40, 50]

for i in range(len(lista) - 1):
    print(lista[i] + lista[i + 1])
````

Conclusão:

> Usar `try/except` para testar possibilidades pode ser útil, mas não deve substituir uma lógica simples e clara.

No caso da lista, se sabemos que queremos parar antes do último elemento, é melhor ajustar o `range`.

# 18. Quando esse estilo faz sentido?

O estilo “tenta fazer e trata se der erro” pode fazer sentido quando:

- o erro é realmente esperado;
- testar antes seria mais complicado;
- a tentativa é mais simples do que a verificação;
- estamos lidando com arquivos, APIs ou dados externos;
- existe risco de algo mudar entre a verificação e a execução.

Exemplo com arquivo:

```python
try:
    abrir arquivo
except FileNotFoundError:
    avisar que não existe
```

Nesse caso, tentar abrir diretamente é natural.

````python
nome_arquivo = "dados.txt"

try:
    with open(nome_arquivo, "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()

    print(conteudo)

except FileNotFoundError:
    print("O arquivo não foi encontrado.")
````

# 19. Quando vira má prática?

O uso de `try/except` vira má prática quando:

- captura erros demais;
- esconde bugs;
- substitui uma condição simples;
- usa `except:` sem especificar o erro;
- usa `pass` sem motivo;
- deixa o programa silenciosamente errado.

Exemplo ruim:

````python
try:
    resultado = 10 / 0
except:
    pass
````

Esse código é perigoso porque o erro aconteceu, mas foi completamente ignorado.

O programa continua como se nada tivesse acontecido.

Isso pode gerar resultados errados e difíceis de descobrir.

# 20. Evite `except:` genérico

Evite escrever:

```python
except:
    print("Erro")
```

Prefira capturar erros específicos:

```python
except ValueError:
    print("Erro de valor.")
```

ou:

```python
except ZeroDivisionError:
    print("Erro de divisão por zero.")
```

Isso deixa o código mais claro e evita esconder problemas inesperados.

# 21. Exemplo completo: processamento de lista com erros

Imagine que temos uma lista com valores misturados:

```python
valores = ["10", "20", "abc", "30", "x", "40"]
```

Queremos converter para inteiros, ignorando os inválidos.

````python
valores = ["10", "20", "abc", "30", "x", "40"]

validos = []
invalidos = []

for valor in valores:
    try:
        numero = int(valor)
        validos.append(numero)

    except ValueError:
        invalidos.append(valor)

print("Valores válidos:", validos)
print("Valores inválidos:", invalidos)
````

Esse é um uso muito bom de `try/except`.

Não queremos que o programa pare só porque um valor veio errado.  
Queremos processar o que for possível e registrar os problemas.

# 22. Exemplo completo: média de notas

Agora vamos trabalhar com entradas do usuário.

O usuário digita várias notas.  
Quando quiser parar, digita `fim`.

````python
notas = []

while True:
    entrada = input("Digite uma nota ou 'fim' para encerrar: ")

    if entrada.lower() == "fim":
        break

    try:
        nota = float(entrada)
        notas.append(nota)

    except ValueError:
        print("Entrada inválida. Digite um número ou 'fim'.")

if len(notas) > 0:
    media = sum(notas) / len(notas)
    print("Média:", media)
else:
    print("Nenhuma nota válida foi digitada.")
````

Observe que nesse exemplo usamos uma mistura de:

- `if`;
- `while`;
- `try/except`;
- lista;
- conversão de tipos.

Esse é o uso mais realista: tratamento de exceções não aparece sozinho, mas junto da lógica do programa.

# 23. Exemplo completo: mini calculadora

Aqui temos um exemplo que junta:

- entrada do usuário;
- conversão para número;
- escolha de operação;
- divisão por zero;
- mensagens específicas.

````python
try:
    numero1 = float(input("Digite o primeiro número: "))
    numero2 = float(input("Digite o segundo número: "))
    operacao = input("Digite a operação (+, -, *, /): ")

    if operacao == "+":
        resultado = numero1 + numero2
        print("Resultado:", resultado)

    elif operacao == "-":
        resultado = numero1 - numero2
        print("Resultado:", resultado)

    elif operacao == "*":
        resultado = numero1 * numero2
        print("Resultado:", resultado)

    elif operacao == "/":
        resultado = numero1 / numero2
        print("Resultado:", resultado)

    else:
        print("Operação inválida.")

except ValueError:
    print("Erro: digite números válidos.")

except ZeroDivisionError:
    print("Erro: não é possível dividir por zero.")
````

# 24. Boas práticas

Algumas boas práticas importantes:

## 1. Capture erros específicos

Prefira:

```python
except ValueError:
```

em vez de:

```python
except:
```

## 2. Não esconda erros sem motivo

Evite:

```python
except:
    pass
```

## 3. Dê mensagens úteis

Prefira:

```python
print("Digite um número válido.")
```

em vez de:

```python
print("Erro.")
```

## 4. Não coloque código demais dentro do `try`

Coloque dentro do `try` apenas o trecho que realmente pode gerar o erro esperado.

## 5. Use `with` para arquivos

Prefira:

```python
with open(...) as arquivo:
```

em vez de abrir e fechar manualmente.

# 25. Resumo dos principais erros

| Erro | Quando acontece | Exemplo |
|---|---|---|
| `ValueError` | valor inválido para uma conversão/operação | `int("abc")` |
| `ZeroDivisionError` | divisão por zero | `10 / 0` |
| `TypeError` | tipos incompatíveis | `"2" + 2` |
| `IndexError` | índice inexistente em lista | `lista[10]` |
| `KeyError` | chave inexistente em dicionário | `dados["idade"]` |
| `FileNotFoundError` | arquivo inexistente | `open("x.txt")` |
| `json.JSONDecodeError` | JSON mal formatado | `json.load(arquivo)` |

# 26. Resumo da estrutura

## Apenas `try/except`

```python
try:
    código
except AlgumErro:
    tratamento
```

## Vários `except`

```python
try:
    código
except ValueError:
    tratamento
except ZeroDivisionError:
    tratamento
```

## Com `else`

```python
try:
    código
except AlgumErro:
    tratamento
else:
    roda se não houver erro
```

## Com `finally`

```python
try:
    código
except AlgumErro:
    tratamento
finally:
    roda sempre
```

# 27. Conclusão

O `try/except` é uma ferramenta fundamental para escrever programas mais robustos.

Ele não deve ser usado para esconder erros, mas sim para lidar com situações que podem dar errado de forma previsível.

Em resumo:

- use `try` no trecho que pode falhar;
- use `except` para tratar erros específicos;
- use mensagens claras;
- evite capturar tudo genericamente;
- não use `pass` para esconder problemas;
- use `finally` quando precisar garantir uma ação final;
- lembre que, em muitos casos, um simples `if` pode ser mais claro.

A ideia principal é:

> Um bom programa não é aquele que nunca encontra erros.  
> É aquele que sabe lidar bem com os erros quando eles aparecem.
