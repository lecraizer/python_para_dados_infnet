# Web Services e IA com POST

Headers, Body e APIs de IA.

## Headers mais comuns

**Authorization** → Quem sou eu?

```python
{"Authorization": "Bearer TOKEN"}
```

**Content-Type** → O que estou enviando?

```python
{"Content-Type": "application/json"}
```

**Accept** → O que quero receber?

```python
{"Accept": "application/json"}
```

**User-Agent** → Quem está fazendo a requisição?

```python
{"User-Agent": "Mozilla/5.0"}
```


## O que é o Body?

O body contém os dados principais da operação.

Exemplo:

```python
body = {
    "nome": "Luis",
    "idade": 37
}
```

O servidor define quais campos espera receber.

## APIs de IA usam POST

O prompt vai dentro do body, não da URL.

```text
Prompt
↓
Body
↓
POST
```


## O que é OpenRouter?

Não é um modelo.

É um intermediário que permite acessar diversos modelos por uma única API.


````python
import requests

chave_api = "SUA_CHAVE"

url = "https://openrouter.ai/api/v1/chat/completions"

pergunta = "Explique em uma frase o que é uma API."

headers = {
    "Authorization": f"Bearer {chave_api}",
    "Content-Type": "application/json"
}

body = {
    "model": "openrouter/free",
    "messages": [
        {
            "role": "user",
            "content": pergunta
        }
    ]
}

resposta = requests.post(
    url,
    headers=headers,
    json=body
)

dados = resposta.json()
dados["choices"][0]["message"]["content"]
````

SDK:

