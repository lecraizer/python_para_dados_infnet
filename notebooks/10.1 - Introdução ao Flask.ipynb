# Introdução ao Flask e APIs Web

Nesta aula veremos como utilizar o Flask para criar serviços web simples em Python.

O objetivo não é desenvolver páginas web completas, mas entender como programas podem disponibilizar dados através da internet para serem consumidos por navegadores, aplicativos e outros sistemas.

## O que é uma aplicação web?

Quando acessamos um site, normalmente pensamos apenas na interface visual: menus, botões, formulários e imagens.

No entanto, por trás dessa interface existe um sistema responsável por processar solicitações e fornecer informações. Esse sistema é chamado de backend.

De forma simplificada, uma aplicação web pode ser dividida em duas partes:

- Frontend: interface apresentada ao usuário.
- Backend: responsável pela lógica da aplicação e pelos dados.

O frontend envia solicitações ao backend através da internet. O backend processa essas solicitações e devolve respostas.

## Frontend e Backend

Considere um sistema acadêmico.

O frontend pode exibir:

- Nome do aluno
- Curso
- Nota final

Já o backend é responsável por armazenar e fornecer essas informações.

Por exemplo:

### Frontend:
Nome: Ana  
Curso: Python  
Nota: 9.5  

### Backend:  
{  
    "nome": "Ana",  
    "curso": "Python",  
    "nota": 9.5  
}  

Observe que o backend não se preocupa com cores, botões ou layout. Sua função é disponibilizar os dados.

## O que é uma API?

API significa Application Programming Interface.

Uma API é uma forma padronizada de um sistema disponibilizar funcionalidades ou dados para outros sistemas.

Em vez de um usuário acessar diretamente um banco de dados, ele faz uma requisição para uma API.

A API recebe a solicitação, processa a informação necessária e retorna uma resposta.

## Exemplo: ViaCEP

O ViaCEP é uma API pública bastante utilizada.

Ao acessar:

https://viacep.com.br/ws/22041001/json/

recebemos uma resposta semelhante a:

{
    "cep": "22041-001",
    "logradouro": "Avenida Atlântica",
    "bairro": "Copacabana",
    "localidade": "Rio de Janeiro"
}

Observe que não existe uma interface gráfica elaborada.

O objetivo da API é apenas fornecer dados para outros programas.

## Requisições e Respostas

A comunicação entre sistemas normalmente ocorre através do protocolo HTTP.

O processo é simples:

1. Um cliente faz uma requisição.
2. O servidor recebe a requisição.
3. O servidor processa a informação.
4. O servidor devolve uma resposta.

Exemplo:

Navegador
    ↓
Requisição HTTP
    ↓
Servidor Flask
    ↓
Resposta JSON

## O que é Flask?

Flask é um framework Python utilizado para construir aplicações web.

Ele permite transformar funções Python em endereços acessíveis através da internet.

Com poucas linhas de código podemos criar APIs completas.

# Criando nosso primeiro servidor

Vamos criar uma aplicação extremamente simples.


````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Servidor funcionando!"

# app.run(debug=True)
````

# O que é uma rota?

Uma rota representa um endereço da aplicação.

Cada rota é associada a uma função Python.

Por exemplo:

- /
- /aluno
- /curso
- /professor

Quando alguém acessa uma dessas URLs, o Flask executa a função correspondente.


````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Página inicial"

@app.route("/curso")
def curso():
    return "Python para Dados"

@app.route("/professor")
def professor():
    return "Luis Eduardo"

# app.run(debug=True)
````


# Retornando dados em formato JSON

Em APIs, normalmente não retornamos apenas texto.

O formato mais comum é o JSON (*JavaScript Object Notation*).

JSON é uma representação textual de estruturas de dados.


````python
from flask import Flask, jsonify

app = Flask(__name__)

@app.route("/aluno")
def aluno():

    dados = {
        "nome": "Ana",
        "nota": 9.5,
        "curso": "Python"
    }

    return jsonify(dados)

# app.run(debug=True)
````


# O que é jsonify()?

A função `jsonify()` converte automaticamente um dicionário Python para JSON.

Exemplo:

```python
{
    "nome": "Ana",
    "nota": 9.5
}
```

é convertido para:

```json
{
    "nome": "Ana",
    "nota": 9.5
}
```

Essa é a forma mais comum de retornar dados em APIs Flask.
