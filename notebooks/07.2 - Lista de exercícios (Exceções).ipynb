# Exercício 1 — Calculadora segura

Crie um programa que peça ao usuário:

- dois números;
- uma operação matemática: `+`, `-`, `*` ou `/`.

O programa deve calcular o resultado da operação.

Requisitos:

- trate entradas não numéricas;
- trate divisão por zero;
- trate operações inválidas;
- o programa não deve quebrar caso o usuário digite algo errado.

Exemplo de situação inválida:

```python
Digite o primeiro número: 10
Digite o segundo número: 0
Digite a operação: /
Erro: não é possível dividir por zero.
```

````python
# Sua solução aqui
````

## Gabarito — Exercício 1

````python
try:
    numero1 = float(input("Digite o primeiro número: "))
    numero2 = float(input("Digite o segundo número: "))
    operacao = input("Digite a operação (+, -, *, /): ")

    if operacao == "+":
        resultado = numero1 + numero2
    elif operacao == "-":
        resultado = numero1 - numero2
    elif operacao == "*":
        resultado = numero1 * numero2
    elif operacao == "/":
        resultado = numero1 / numero2
    else:
        print("Erro: operação inválida.")
        resultado = None

    if resultado is not None:
        print("Resultado:", resultado)

except ValueError:
    print("Erro: digite apenas números válidos.")

except ZeroDivisionError:
    print("Erro: não é possível dividir por zero.")
````

## Gabarito — Exercício 2

````python
valores = ["10", "25", "abc", "40", "3.5", "x", "100"]

validos = []
invalidos = []

for valor in valores:
    try:
        numero = int(valor)
        validos.append(numero)
    except ValueError:
        invalidos.append(valor)

print("Valores válidos:", validos)
print("Valores inválidos:", invalidos)
````

# Exercício 3 — Média de notas com entradas inválidas

Crie um programa que peça notas ao usuário até que ele digite `fim`.

Requisitos:

- cada nota deve ser convertida para `float`;
- entradas inválidas devem ser ignoradas com uma mensagem de erro;
- ao final, calcule a média apenas das notas válidas;
- se nenhuma nota válida for digitada, mostre uma mensagem apropriada em vez de deixar o programa quebrar.

Exemplo:

```python
Digite uma nota: 8
Digite uma nota: abc
Entrada inválida.
Digite uma nota: 7
Digite uma nota: fim

Média: 7.5
```

````python
# Sua solução aqui
````

## Gabarito — Exercício 3

````python
notas = []

while True:
    entrada = input("Digite uma nota ou 'fim' para encerrar: ")

    if entrada.lower() == "fim":
        break

    try:
        nota = float(entrada)
        notas.append(nota)
    except ValueError:
        print("Entrada inválida. Digite um número ou 'fim'.")

try:
    media = sum(notas) / len(notas)
    print("Média:", media)
except ZeroDivisionError:
    print("Nenhuma nota válida foi digitada.")
````

## Gabarito — Exercício 7

````python
import json

try:
    with open("dados.json", "r", encoding="utf-8") as arquivo:
        dados = json.load(arquivo)

    print("Dados carregados com sucesso:")
    print(dados)

except FileNotFoundError:
    print("Erro: o arquivo dados.json não foi encontrado.")

except json.JSONDecodeError:
    print("Erro: o arquivo existe, mas não contém um JSON válido.")

except OSError:
    print("Erro: não foi possível abrir ou ler o arquivo.")
````

## Gabarito — Exercício 8

````python
compras = [
    {"produto": "camisa", "preco": "50", "quantidade": "2"},
    {"produto": "calça", "preco": "abc", "quantidade": "1"},
    {"produto": "tênis", "preco": "200", "quantidade": "x"},
    {"produto": "boné", "preco": "40", "quantidade": "3"}
]

total_geral = 0
produtos_com_erro = []

for compra in compras:
    try:
        produto = compra["produto"]
        preco = float(compra["preco"])
        quantidade = int(compra["quantidade"])

        total = preco * quantidade
        total_geral += total

        print(f"{produto}: R$ {total:.2f}")

    except ValueError:
        produtos_com_erro.append(compra.get("produto", "produto desconhecido"))

print("Total geral:", total_geral)
print("Produtos com erro:", produtos_com_erro)
````

# Exercício 9 — Mini sistema de saque

Crie um programa que simule um saque bancário.

Use um saldo inicial, por exemplo:

```python
saldo = 500
```

O usuário deve digitar o valor que deseja sacar.

Requisitos:

- trate entradas não numéricas;
- não permita saque negativo;
- não permita saque maior que o saldo;
- se o saque for válido, atualize o saldo;
- permita várias operações até o usuário digitar `sair`.

Observação: neste exercício, não crie exceções personalizadas. Use condicionais junto com `try/except`.

````python
saldo = 500

# Sua solução aqui
````

## Gabarito — Exercício 9

````python
saldo = 500

while True:
    entrada = input("Digite o valor do saque ou 'sair': ")

    if entrada.lower() == "sair":
        break

    try:
        valor = float(entrada)

        if valor <= 0:
            print("Erro: o valor do saque deve ser positivo.")
        elif valor > saldo:
            print("Erro: saldo insuficiente.")
        else:
            saldo -= valor
            print(f"Saque realizado. Saldo atual: R$ {saldo:.2f}")

    except ValueError:
        print("Erro: digite um valor numérico válido.")
````

# Exercício 10 — Interpretador simples de comandos

Crie um programa que leia comandos digitados pelo usuário no formato:

```python
soma 10 20
subtrai 30 5
multiplica 4 8
divide 10 2
```

O programa deve interpretar o comando e executar a operação.

Requisitos:

- trate comandos inexistentes;
- trate quantidade errada de argumentos;
- trate valores não numéricos;
- trate divisão por zero;
- o programa deve continuar rodando até o usuário digitar `sair`.

Exemplos de erros:

```python
soma 10
Erro: quantidade incorreta de argumentos.

divide 10 0
Erro: divisão por zero.

potencia 2 3
Erro: comando desconhecido.
```

````python
# Sua solução aqui
````

## Gabarito — Exercício 10

````python
while True:
    entrada = input("Digite um comando ou 'sair': ")

    if entrada.lower() == "sair":
        break

    partes = entrada.split()

    try:
        if len(partes) != 3:
            print("Erro: quantidade incorreta de argumentos.")
            continue

        comando = partes[0]
        numero1 = float(partes[1])
        numero2 = float(partes[2])

        if comando == "soma":
            resultado = numero1 + numero2
        elif comando == "subtrai":
            resultado = numero1 - numero2
        elif comando == "multiplica":
            resultado = numero1 * numero2
        elif comando == "divide":
            resultado = numero1 / numero2
        else:
            print("Erro: comando desconhecido.")
            continue

        print("Resultado:", resultado)

    except ValueError:
        print("Erro: os argumentos devem ser números.")

    except ZeroDivisionError:
        print("Erro: divisão por zero.")
````

# Gabarito — Perguntas finais

## 1.

O `if` é usado para testar condições esperadas do programa.  
O `try/except` é usado para lidar com erros que podem acontecer durante a execução.

Exemplo:

- `if` pode verificar se uma operação é permitida.
- `try/except` pode capturar um erro de conversão ao tentar transformar `"abc"` em número.

## 2.

Porque `except:` captura qualquer erro, inclusive erros que talvez sejam bugs do código.  
Isso pode esconder problemas importantes e dificultar a depuração.

É melhor capturar erros específicos, como:

```python
except ValueError:
```

ou:

```python
except FileNotFoundError:
```

## 3.

Faz sentido deixar o programa quebrar quando o erro indica um problema real no código ou uma situação que não deveria acontecer.

Por exemplo, durante o desenvolvimento, esconder todos os erros pode atrapalhar, porque o programador deixa de perceber bugs.

## 4.

Para o usuário final, normalmente é melhor mostrar uma mensagem amigável.

Exemplo:

```python
Erro: digite um número válido.
```

é melhor do que:

```python
ValueError: could not convert string to float
```

Mas, para o programador, a mensagem técnica pode ser útil durante a depuração.

## 5.

O `finally` não é obrigatório.

Ele é útil quando queremos executar alguma ação independentemente de ter ocorrido erro ou não.

Exemplos:

- fechar arquivo;
- encerrar conexão;
- mostrar uma mensagem final;
- liberar algum recurso.
