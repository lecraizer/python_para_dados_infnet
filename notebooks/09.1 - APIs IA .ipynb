# 1. O que é uma API de IA?

Até agora utilizamos APIs como:

```text
ViaCEP
↓
CEP
↓
JSON
```

Uma API de IA funciona de forma parecida:

```text
Prompt (texto)
↓
Modelo
↓
Resposta (texto)
```

Ou seja, enviamos uma string e recebemos outra string.

# 3. Primeiro Exemplo - Gemini

````python
from google import genai

api_key = "SUA_CHAVE_AQUI"

client = genai.Client(
    api_key=api_key
)

prompt = "Explique Machine Learning em uma frase."

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
````

# 4. Primeiro Exemplo - OpenAI

````python
from openai import OpenAI

api_key = "SUA_CHAVE_AQUI"

client = OpenAI(
    api_key=api_key
)

prompt = "Explique Machine Learning em uma frase."

response = client.responses.create(
    model="gpt-4.1-mini",
    input=prompt
)

print(response.output_text)
````

# 5. Temperature

Temperature controla a criatividade da IA.

## Temperature baixa

Mais previsível.

```python
temperature=0
```

## Temperature alta

Mais criativa.

```python
temperature=1
```


````python
prompt = "Me dê um nome para uma banda de rock."

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt,
    config={
        "temperature": 0
    }
)

print(response.text)
````

````python
prompt = "Me dê um nome para uma banda de rock."

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt,
    config={
        "temperature": 1
    }
)

print(response.text)
````

# 6. Max Output Tokens

Controla o tamanho máximo da resposta.


````python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Explique redes neurais.",
    config={
        "max_output_tokens": 50
    }
)

print(response.text)
````

# 8. Respostas Estruturadas em JSON

A IA pode gerar JSON para ser processado pelo Python.


````python
import json

prompt = '''
Retorne apenas JSON:

{
    "nome": "",
    "idade": 0,
    "profissao": ""
}

Albert Einstein
'''

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
````

````python
dados = json.loads(response.text)

print(dados["nome"])
print(dados["idade"])
print(dados["profissao"])
````

# 9. Criando uma Função Auxiliar

````python
def perguntar(prompt):

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text


resposta = perguntar(
    "Explique overfitting em uma frase."
)

print(resposta)
````

# Complemento: Fazendo chamadas HTTP puras com POST

Até agora utilizamos os SDKs da OpenAI e do Gemini.

Mas por trás dos SDKs existe uma requisição HTTP parecida com:

```python
requests.post(
    url,
    headers=headers,
    json=body
)
```

O objetivo desta seção é mostrar os bastidores da comunicação.


````python

import requests

url = "https://api.openai.com/v1/responses"

headers = {
    "Authorization": f"Bearer {openai_api_key}",
    "Content-Type": "application/json"
}

body = {
    "model": "gpt-4.1-mini",
    "input": "Explique Machine Learning em uma frase."
}

resposta = requests.post(
    url,
    headers=headers,
    json=body
)

dados = resposta.json()

print(dados)
````

````python

import requests

url = (
    "https://generativelanguage.googleapis.com/v1beta/"
    f"models/gemini-1.5-flash:generateContent?key={gemini_api_key}"
)

headers = {
    "Content-Type": "application/json"
}

body = {
    "contents": [
        {
            "parts": [
                {
                    "text": "Explique Machine Learning em uma frase."
                }
            ]
        }
    ]
}

resposta = requests.post(
    url,
    headers=headers,
    json=body
)

dados = resposta.json()

print(dados)
````


## Observação

A resposta da IA não vem diretamente como string.

Ela vem dentro de um JSON.

OpenAI:

```python
dados["output"][0]["content"][0]["text"]
```

Gemini:

```python
dados["candidates"][0]["content"]["parts"][0]["text"]
```

Os SDKs escondem essa complexidade e retornam algo mais simples, como:

```python
response.output_text
```

ou

```python
response.text
```
