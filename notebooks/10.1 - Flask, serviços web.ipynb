# 2. Primeiro servidor Flask

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():

    return "Servidor funcionando!"

app.run(debug=True)
````

# 3. Entendendo rotas

Uma rota é uma URL que executa uma função Python.

Exemplo:

```python
@app.route("/idade")
```

vira:

```text
http://127.0.0.1:5000/idade
```

````python
from flask import Flask

app = Flask(__name__)

@app.route("/idade")
def idade():

    return "35"

# app.run(debug=True)
````

# 4. Criando múltiplas rotas

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():

    return "Página inicial"

@app.route("/curso")
def curso():

    return "Python para Dados"

@app.route("/professor")
def professor():

    return "Luis"

# app.run(debug=True)
````

# 5. Rotas com parâmetros

````python
from flask import Flask

app = Flask(__name__)

@app.route("/quadrado/<int:x>")
def quadrado(x):

    return str(x**2)

# app.run(debug=True)
````

Se acessar:

```text
http://127.0.0.1:5000/quadrado/7
```

Retorna:

```text
49
```

# 6. Retornando JSON

APIs normalmente devolvem JSON.

Flask possui:

```python
jsonify()
```

````python
from flask import Flask
from flask import jsonify

app = Flask(__name__)

@app.route("/aluno")
def aluno():

    dados = {
        "nome":"Ana",
        "nota":9.5,
        "curso":"Python"
    }

    return jsonify(dados)

# app.run(debug=True)
````

Se abrir:

```text
/aluno
```

Receberá JSON.

# 7. Lista JSON

````python
from flask import Flask
from flask import jsonify

app = Flask(__name__)

@app.route("/usuarios")
def usuarios():

    dados = [

        {
            "nome":"Ana",
            "idade":20
        },

        {
            "nome":"Carlos",
            "idade":32
        }

    ]

    return jsonify(dados)

# app.run(debug=True)
````

# 8. Consumindo a própria API

Agora vamos ligar Flask + requests.

````python
import requests

resp = requests.get(
    "http://127.0.0.1:5000/aluno"
)

resp.json()
````

Fluxo:

```text
Python cliente

↓

requests.get()

↓

Flask local

↓

JSON

↓

Python cliente
```

# 9. Criando endpoint de notas

````python
from flask import Flask
from flask import jsonify

app = Flask(__name__)

@app.route("/notas")
def notas():

    dados = {

        "Maria":8.5,
        "Pedro":7.0,
        "Lucas":9.8

    }

    return jsonify(dados)

# app.run(debug=True)
````

# 10. Consumindo endpoint local

````python
resp = requests.get(
    "http://127.0.0.1:5000/notas"
)

dados = resp.json()

dados
````

````python
for nome, nota in dados.items():

    print(nome, nota)
````

# 11. Endpoint com parâmetros

````python
from flask import Flask

app = Flask(__name__)

@app.route("/dobro/<int:x>")
def dobro(x):

    return str(
        x * 2
    )

# app.run(debug=True)
````

Teste:

```text
/dobro/30
```

# 12. Mini projeto guiado

Sistema local de produtos.

````python
from flask import Flask
from flask import jsonify

app = Flask(__name__)

produtos = [

    {
        "id":1,
        "nome":"Notebook",
        "preco":5500
    },

    {
        "id":2,
        "nome":"Mouse",
        "preco":150
    }

]

@app.route("/produtos")
def listar_produtos():

    return jsonify(produtos)

# app.run(debug=True)
````

Cliente:

````python
resp = requests.get(
    "http://127.0.0.1:5000/produtos"
)

produtos = resp.json()

produtos
````

````python
for produto in produtos:

    print(
        produto["nome"],
        produto["preco"]
    )
````
