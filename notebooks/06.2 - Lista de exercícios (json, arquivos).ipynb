## Exercício 1 — Temperaturas da semana

Crie um programa que leia `temperaturas.json` e gere `relatorio_temperaturas.json` contendo média, maior e menor temperatura.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json

with open("temperaturas.json", "r", encoding="utf-8") as f:
    dados = json.load(f)

temperaturas = dados["temperaturas"]

soma = 0
maior = temperaturas[0]
menor = temperaturas[0]

for temp in temperaturas:
    soma += temp

    if temp > maior:
        maior = temp

    if temp < menor:
        menor = temp

media = soma / len(temperaturas)

relatorio = {
    "cidade": dados["cidade"],
    "media": round(media, 2),
    "maior": maior,
    "menor": menor
}

with open("relatorio_temperaturas.json", "w", encoding="utf-8") as f:
    json.dump(relatorio, f, indent=2, ensure_ascii=False)
```

## Exercício 2 — Clientes inativos

Leia `clientes.json` e gere `clientes_inativos.json` contendo apenas clientes sem compras.


```python
# Sua resposta aqui
```

### Gabarito

## Exercício 3 — TXT para JSON

Leia `alunos.txt` e gere `alunos.json`.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json

alunos = []

with open("alunos.txt", "r", encoding="utf-8") as f:
    linhas = f.readlines()

for linha in linhas:
    linha = linha.strip()

    partes = linha.split(";")

    nome = partes[0]
    nota = float(partes[1])

    aluno = {
        "nome": nome,
        "nota": nota
    }

    alunos.append(aluno)

with open("alunos.json", "w", encoding="utf-8") as f:
    json.dump(alunos, f, indent=2, ensure_ascii=False)
```

## Exercício 4 — Filmes acima da média

Leia `filmes.json` e gere `melhores_filmes.json` contendo apenas filmes acima da média.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json

with open("filmes.json", "r", encoding="utf-8") as f:
    filmes = json.load(f)

soma = 0

for filme in filmes:
    soma += filme["nota"]

media = soma / len(filmes)

melhores = []

for filme in filmes:
    if filme["nota"] > media:
        melhores.append(filme)

with open("melhores_filmes.json", "w", encoding="utf-8") as f:
    json.dump(melhores, f, indent=2, ensure_ascii=False)

print(f"Média geral: {media:.2f}")
```

## Exercício 7 — Pedidos com itens aninhados

Leia `pedidos.json` e gere `resumo_pedidos.json` com o total gasto por cliente.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json

with open("pedidos.json", "r", encoding="utf-8") as f:
    pedidos = json.load(f)

resumo = []

for pedido in pedidos:

    total = 0

    for item in pedido["itens"]:
        total += item["valor"]

    resumo.append({
        "cliente": pedido["cliente"],
        "total": total
    })

with open("resumo_pedidos.json", "w", encoding="utf-8") as f:
    json.dump(resumo, f, indent=2, ensure_ascii=False)
```

## Exercício 8 — Sorteio de senhas temporárias

Use `random` para gerar 15 senhas temporárias de 6 caracteres e grave `senhas.json`.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json
import random
import string

senhas = []

caracteres = string.ascii_uppercase + string.digits

for i in range(15):

    senha = ""

    for j in range(6):
        senha += random.choice(caracteres)

    senhas.append(senha)

resultado = {
    "quantidade": len(senhas),
    "senhas": senhas
}

with open("senhas.json", "w", encoding="utf-8") as f:
    json.dump(resultado, f, indent=2, ensure_ascii=False)

print("Arquivo senhas.json criado.")
```

## Exercício 9 — Juntando dois CSVs

Leia `clientes.csv` e `compras.csv` e gere `resumo_clientes.json` com o total gasto por cliente.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json

clientes = {}

with open("clientes.csv", "r", encoding="utf-8") as f:

    linhas = f.readlines()

    for linha in linhas[1:]:

        linha = linha.strip()

        partes = linha.split(",")

        id_cliente = partes[0]
        nome = partes[1]

        clientes[id_cliente] = {
            "nome": nome,
            "total": 0
        }

with open("compras.csv", "r", encoding="utf-8") as f:

    linhas = f.readlines()

    for linha in linhas[1:]:

        linha = linha.strip()

        partes = linha.split(",")

        id_cliente = partes[0]
        valor = float(partes[1])

        clientes[id_cliente]["total"] += valor

resumo = []

for id_cliente in clientes:

    resumo.append({
        "nome": clientes[id_cliente]["nome"],
        "total": round(clientes[id_cliente]["total"], 2)
    })

with open("resumo_clientes.json", "w", encoding="utf-8") as f:
    json.dump(resumo, f, indent=2, ensure_ascii=False)
```

## Exercício 10 — Cruzando dois JSONs

Leia `produtos.json` e `vendas.json` e gere `vendas_produtos.json`.


```python
# Sua resposta aqui
```

### Gabarito

```python
import json

with open("produtos.json", "r", encoding="utf-8") as f:
    produtos = json.load(f)

with open("vendas.json", "r", encoding="utf-8") as f:
    vendas = json.load(f)

produtos_por_id = {}

for produto in produtos:

    id_produto = produto["id"]

    produtos_por_id[id_produto] = {
        "nome": produto["nome"],
        "categoria": produto["categoria"],
        "quantidade_total": 0
    }

for venda in vendas:

    id_produto = venda["id_produto"]
    quantidade = venda["quantidade"]

    produtos_por_id[id_produto]["quantidade_total"] += quantidade

resultado = []

for id_produto in produtos_por_id:

    resultado.append({
        "produto": produtos_por_id[id_produto]["nome"],
        "categoria": produtos_por_id[id_produto]["categoria"],
        "quantidade_total": produtos_por_id[id_produto]["quantidade_total"]
    })

with open("vendas_produtos.json", "w", encoding="utf-8") as f:
    json.dump(resultado, f, indent=2, ensure_ascii=False)
```
