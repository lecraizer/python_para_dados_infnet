# Raise e Assert em Python

Notebook curto complementar sobre `raise` e `assert`.

## Raise

`raise` serve para gerar um erro propositalmente.

````python
def sacar(valor):

    if valor < 0:

        raise ValueError(
            "Valor inválido."
        )

    print("Saque realizado.")

sacar(50)
````

Teste depois:

```python
sacar(-10)
```

````python
def cadastrar_idade(idade):

    if idade < 0:

        raise ValueError(
            "Idade inválida."
        )

    if idade > 120:

        raise ValueError(
            "Idade muito alta."
        )

    print("Cadastro OK")

cadastrar_idade(35)
````

## Assert

`assert` verifica algo que deveria ser verdadeiro.

````python
idade = 20

assert idade > 0

print(
    "Tudo certo."
)
````

Teste depois:

```python
idade=-5
assert idade>0
```

````python
def calcular_media(notas):

    assert len(notas) > 0

    return sum(notas)/len(notas)

print(
    calcular_media(
        [8,9,10]
    )
)
````

## Comparação

| Recurso | Ideia |
|---|---|
| raise | criar erro |
| assert | verificar condição |

Resumo:

```text
raise

↓

"entrada inválida"

assert

↓

"isso deveria ser verdade"
```
